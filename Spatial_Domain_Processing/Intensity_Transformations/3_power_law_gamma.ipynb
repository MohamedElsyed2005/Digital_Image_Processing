### Power-Law (Gamma) Transformations

Form:
$$ s = c \; r^\gamma$$

where: 
- $$ c,\; \gamma$$
are positive consants.

  <img src="Images/3_0_image.png" alt="gamma curves" width="350"/>
  <figcaption>Gamma curves.</figcaption>
</figure>

> As with `log Transformations`, `power-law` curves with fractional values of **gamma** map a narrow range of dark input values into a wider range of output values, 
>with the opposite being true for higher values of input levels.

**Gamma correction** is important for displaying images on a screen correctly, to prevent bleaching or darkening of images when viewed from different types of monitors with different display settings. 

This is done because our eyes perceive images in a gamma-shaped curve, whereas cameras capture images in a linear fashion.

```python
import cv2 
import matplotlib.pyplot as plt

image = cv2.imread('Images/3_1_image.png', cv2.IMREAD_GRAYSCALE)

plt.figure(figsize= (10,6))
plt.imshow(image, cmap = "gray")
plt.axis("off")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

gammas = [0.1, 0.5, 1.2, 2.2]

plt.figure(figsize=(18, 5)) 

for i, gamma in enumerate(gammas):
    # Apply gamma correction
    gamma_corrected = np.array(255 * (image / 255) ** gamma, dtype='uint8')

    plt.subplot(1, len(gammas), i + 1)
    plt.imshow(gamma_corrected, cmap="gray")
    plt.title(f'Gamma = {gamma}')
    plt.axis('off')

plt.tight_layout()
plt.show()
```

### Effect of Gamma Correction

As can be observed from both the output images and the corresponding graph:

- **γ > 1** (represented by the curve labeled *“nth power”*):  
  The pixel intensities decrease, causing the image to become **darker**.

- **γ < 1** (represented by the curve labeled *“nth root”*):  
  The pixel intensities increase, causing the image to become **lighter**.

This behavior illustrates how gamma correction controls image brightness by applying a nonlinear transformation to pixel intensities.
