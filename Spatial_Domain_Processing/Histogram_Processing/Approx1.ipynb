### Transformation of Random Variables

Here we study how the distribution of a random variable changes when the variable is transformed in a deterministic way.

$$ s = T(r) $$

We have a random variable `r` taking values in set $A$, and a function $T: A \to B$. Then $s = T(r)$ is a new random variable taking values in $B$.

If the distribution of `r` is known, how do we find the distribution of `s`?

But first recall that for $C \subseteq B$:

$$T^{-1}(C) = \{r \in A: T(r) \in C\}$$

However, frequently the distribution of `r` is known either through its distribution function `F` or its pdf `f`, and we would similarly like to find the distribution function or probability density function of `s`.

This is a difficult problem in general, because as we will see, even simple transformations of variables with simple distributions can lead to variables with complex distributions.

---

> **The change of variables formula**

When the transformation `T` is **one-to-one** (strictly monotonic increasing) and smooth, there is a formula for the probability density function (pdf) of `s`. This is known as the **change of variables formula**.

**Since `T` is one-to-one, it has an inverse function $T^{-1}$**

Let:
- $G(s)$ → the distribution function of `s`
- $g(s)$ → the pdf of `s`

Suppose that `T` is strictly increasing on `A`. For $s \in B$:

1. $$G(s) = P(S \leq s) = P(T(r) \leq s) = P(r \leq T^{-1}(s))$$

   $$\because s = T(r) \;\; \therefore r = T^{-1}(s)$$

2. $$\therefore G(s) = P(S \leq s) = P(T(r) \leq s) = P(r \leq T^{-1}(s)) = F(T^{-1}(s))$$

3. $$g(s) = \frac{dG(s)}{ds} = \frac{d}{ds} F(T^{-1}(s)) = F'(T^{-1}(s)) \cdot \frac{d}{ds} T^{-1}(s) = f(T^{-1}(s)) \cdot \frac{d}{ds} T^{-1}(s)$$

   > **Note:** $F' = f$ (the derivative of the CDF is the PDF)

If `T` is strictly increasing on `A`, then the probability density function (pdf) $g$ of $s$ is given by:

$$
g(s) = f(T^{-1}(s)) \cdot \frac{d}{ds} T^{-1}(s)
$$

$$
\because s = T(r) \;\; \text{and} \;\; r = T^{-1}(s)
$$

$$
\therefore g(s) = f(r) \cdot \frac{dr}{ds}
$$

### Matching with the Standard Formula

To match with the standard histogram equalization formula, let:

1. $P_s(s) = g(s)$ → PDF of the output variable
2. $P_r(r) = f(r)$ → PDF of the input variable

Therefore:

$$
P_s(s) = P_r(r)\left|\frac{dr}{ds}\right|
$$

> **Note:** The absolute value $\left|\frac{dr}{ds}\right|$ ensures the PDF remains positive for both increasing and decreasing transformations.
